Splitting the Data into Training data & Test Data

```python
# Drop rows with any missing values
heart_data = heart_data.dropna()
```

```python
X = heart_data.drop(columns='Heart_Disease', axis=1)
Y = heart_data['Heart_Disease']

# Split the cleaned data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=2)

# Print the shapes to verify
print(X.shape, X_train.shape, X_test.shape)
```

```python
X = heart_data.drop(columns='Heart_Disease', axis=1)
Y = heart_data['Heart_Disease']

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=2)
```

```python
print('X shape:', X.shape)
print('X_train shape:', X_train.shape)
print('X_test shape:', X_test.shape)
```

Model Training

Logistic Regression

```python
model = LogisticRegression()
```

```python
# Drop non-numeric identifier and encode categorical columns
X_train_prediction = pd.get_dummies(X_train.drop(columns=['Patient_ID']), drop_first=True)
X_test_accuracy = pd.get_dummies(X_test.drop(columns=['Patient_ID']), drop_first=True)

# Ensure both train and test have the exact same columns after encoding
X_train_prediction, X_test_accuracy = X_train_prediction.align(X_test_accuracy, join='left', axis=1, fill_value=0)

# Initialize and train the Logistic Regression model
model = LogisticRegression(max_iter=1000)
model.fit(X_train_prediction, Y_train)
print("Model trained successfully!")
```

```python
# Use the correctly-shaped X_train_prediction (the 2D training feature matrix) to make predictions
train_predictions = model.predict(X_train_prediction)
training_data_accuracy = accuracy_score(Y_train, train_predictions)
print('Accuracy on Training data: ', training_data_accuracy)
```

```python
# accuracy on test data
test_predictions = model.predict(X_test_accuracy)
test_data_accuracy = accuracy_score(Y_test, test_predictions)
print('Accuracy on Test data: ', test_data_accuracy)
```
